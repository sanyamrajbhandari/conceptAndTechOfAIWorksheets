```python
import numpy as np

x = np.array([[1,2],
              [3,5]])
y = np.array([[5,6],
              [7,8]])

v = np.array([9,10])
w = np.array([11,12])

print("Add x and y: ",np.add(x,y))
print("Add v and w: ",np.add(v,w))

print("Subtract y from x: ",np.subtract(x,y))
print("Subtract w from v: ",np.subtract(v,w))

#3 Multiply the arrays with any integers of choice
print("Multiplying x by 5: ",np.multiply(x,5))
print("Multiplying y by 5: ",np.multiply(y,5))
print("Multiplying v by 5: ",np.multiply(v,5))
print("Multiplying w by 5: ",np.multiply(w,5))

#4 FInd the square of each element of the array
print("The square of the elements of x are: ",(np.power(x,2)))
print("The square of the elements of y are: ",(np.power(y,2)))
print("The square of the elements of v are: ",(np.power(v,2)))
print("The square of the elements of w are: ",(np.power(w,2)))

#5 Find the dot product between: v(and)w ; x(and)v ; x(and)y.

print("The dot product between v and w is ", np.dot(v,w))
print("The dot product between x and v is ", np.dot(x,v))
print("The dot product between x and y is ", np.dot(x,y))

#6. Concatenate x(and)y along row and Concatenate v(and)w along column.

# Concatenating x and y along row (axis=0 or vstack)
np.concatenate((x,y),axis=0)
np.vstack((x,y))

# Concatenating v(and) w along the column (axis =1) or h stack
np.concatenate((v,w),axis=0) # they are 1D arrays so they only have one axis(axis 0)
np.hstack((v,w))

#Concatenate x(and)v; if you get an error, observe and explain why did you get the error?
np.vstack((x,v))
# np.hstack((x,v)) we get an error while doing this because  the number of rows don't match


```

```python
# AB  != BA


A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

ab = np.dot(A,B)
ba = np.dot(B,A)
if(np.array_equal(ab,ba)):
  print("AB is equal to BA")
else:
  print("AB is not equal to BA")
```

```python
#Prove (AB)Transpose = B Transpose A Transpose

A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

ab = np.dot(A,B)
abTranspose = ab.T

bTranspose = B.T
aTranspose = A.T

dotBtAt = np.dot(bTranspose,aTranspose)

if(np.array_equal(abTranspose,dotBtAt)):
    print(True)

```

```python
'''
Solve the following system of Linear equation using Inverse Methods.

2x − 3y + z = −1
x − y + 2z = −3
3x + y − z = 9
'''

A = np.array([[2,-3,1],
             [1,-1,2],
             [3,1,-1]])
B = np.array([[-1],
             [-3],
             [9]])

A_Inverse = np.linalg.inv(A)
X = np.dot(A_Inverse,B)
print(X)
```
