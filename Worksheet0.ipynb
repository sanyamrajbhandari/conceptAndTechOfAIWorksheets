## **Task 1**

```python
import matplotlib.pyplot as plt


time_data = [
(3.5, 2.0, 7.0), (5.0, 1.5, 6.5), (2.5, 3.0, 8.0),
(4.0, 2.0, 6.0), (1.5, 4.5, 9.0), (3.0, 2.5, 7.5),
(5.5, 1.0, 6.0), (2.0, 3.5, 8.5), (4.5, 2.0, 7.0),
(3.0, 3.0, 7.5), (6.0, 1.5, 6.0), (2.5, 4.0, 8.0),
(4.0, 2.5, 7.0), (5.0, 2.0, 6.5), (3.5, 2.5, 7.0)
]

low = [] #less than 3 hours
moderate = [] # 3 to 5 hours
high = [] # More than 5 hours

for times in time_data:
  if times[0]>5.0:
    high.append(times[0])
  elif times[0]>=3.0:
    moderate.append(times[0])
  elif times[0]<3.0:
    low.append(times[0])
  else:
    print('Invalid Data set')

print(low)
print(moderate)
print(high)

#Study Time Number
lowStudyTimeDays = len(low)
moderateStudyTimeDays = len(moderate)
highStudyTimeDays = len(high)

print(f"The number of low study time days were {lowStudyTimeDays}")
print(f"The number of moderate study time days were {moderateStudyTimeDays}")
print(f"The number of low study time days were {highStudyTimeDays}")

#Task 3 Convert Hours into minutes

study_minutes = []
for times in time_data:
  study_minutes.append(int(times[0]*60))
print(study_minutes)


#Task 4
study_hours = []
entertainment_hours = []
sleep_hours = []

for times in time_data:
  study_hours.append(times[0])
  entertainment_hours.append(times[1])
  sleep_hours.append(times[2])

average_study_hours = sum(study_hours)/len(study_hours)
average_entertainment_hours = sum(entertainment_hours)/len(entertainment_hours)
average_sleep_hours = sum(sleep_hours)/len(sleep_hours)

print(average_study_hours)
print(average_entertainment_hours)
print(average_sleep_hours)

#Task 5
import matplotlib.pyplot as plt

study_hours = [t[0] for t in time_data]
sleep_hours = [t[2] for t in time_data]
plt.scatter(study_hours, sleep_hours)
plt.xlabel("Study Hours")
plt.ylabel("Sleep Hours")
plt.title("Study Hours vs Sleep Hours")
plt.grid(True)
plt.show()
```

# Recursive Tasks


```python
#task 1

nested_list = [1, [2, [3, 4], 5], 6, [7, 8]]

def sum_nested_list(nested_list):
  total = 0

  for item in nested_list :
    if isinstance(item,list):
      total+= sum_nested_list(item)
    else:
      total+= item
  return total
print(sum_nested_list(nested_list))
```
