# Exploratory Data Analysis Practice Exercise-1

## Which passenger had the highest fare paid relative to their age?

1. Add a column to the Titanic dataset, fare per year, containing the fare divided by the age of
the passenger(i.e., Fare/Age).

```python
import pandas as pd

titanic_df = pd.read_csv('/content/drive/MyDrive/Datasets/Week2/Titanic-Dataset.csv')

#Filling the empty values of age with the mean to aid in division
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].mean())

titanic_df['fare_per_year'] = titanic_df['Fare']/titanic_df['Age']
titanic_df.head()
```

2. Subset rows where fare_per_year is higher than 5, assigning this to high_fare_age.

```python
high_fare_age = titanic_df.loc[titanic_df['fare_per_year']>5]
high_fare_age.head()
```

3. Sort high_fare_age by descending fare per year, assigning this to high fare age srt.

```python
high_fare_age_srt = high_fare_age.sort_values(by='fare_per_year', ascending = False)
display(high_fare_age_srt.head())
print()
print("Tail")
print()

high_fare_age_srt.tail()
```

4. Select only the Name and fare_per_yearcolumns of high fare age srt and save the result as result.

```python
result = high_fare_age_srt[['Name','fare_per_year']]
result
```

From the result we can observe that Passenger Allison, Master. Hudson Trevor had the highest fare paid relative to their age

# Exploratory Data Analysis with Group-by Method Practice Exercise:

## What percent of the total fare revenue came from each passenger class?

1. Calculate the total Fare paid across all passengers in the Titanic dataset.

```python
import pandas as pd

titanic_df = pd.read_csv('/content/drive/MyDrive/Datasets/Week2/Titanic-Dataset.csv')

total_fare = titanic_df['Fare'].sum()
print(total_fare)
```

2. Subset for passengers in first class (Pclass is 1) and calculate their total fare.
3. Do the same for second class (Pclass is 2) and third class (Pclass is 3).
4. Combine the fare totals from first, second, and third classes into a list.

```python
grouped_fares = titanic_df.groupby('Pclass')['Fare'].sum()
display(grouped_fares)
print("Proportion fare revenue by class is given below!")
proportion_fare_revenue_by_class = grouped_fares/total_fare
proportion_fare_revenue_by_class
```

## What percent of the total number of passengers on the Titanic belonged to each age group (e.g., child, adult, senior)?

1. Create a new column, age group, that categorizes passengers into "child" (age < 18), "adult" (age
18-64), and "senior" (age 65 and above).

```python
# Filling the age to adjust for null values
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].mean())

titanic_df["age_group"] = titanic_df["Age"].apply(
    lambda age: "child" if age < 18 else ("adult" if age < 65 else "senior")
)
titanic_df.tail()
```

2. Calculate the total number of passengers on the Titanic

```python
no_of_passengers = len(titanic_df)
no_of_passengers
```

3. Count the number of passengers in each age group

```python
age_groups = titanic_df.groupby('age_group')['Age'].count()
age_groups.head()
```

4. Divide the count of each age group by the total number of passengers to get the proportion of passengers
in each age group.

```python
age_group_proportion = age_groups/no_of_passengers
age_group_proportion
```

```python
age_group_percentage = age_group_proportion * 100
print("The percentage of people according to age group is:")
age_group_percentage
```
